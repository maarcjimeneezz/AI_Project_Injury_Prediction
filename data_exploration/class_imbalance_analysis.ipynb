# **1. Class Imbalance Analysis**

The objective of this analysis is to evaluate the ratio between the two target labels: 
* **Injured:** $[1]$
* **Non-injured:** $[0]$

### Why this matters:
Due to the physiological nature of sports, injuries are "rare events" compared to the total number of training days. This creates a significant **Class Imbalance**. 

From a machine learning perspective, this means we cannot rely on "standard accuracy" as a performance metric. If $98\%$ of the data consists of non-injury days, an algorithm that blindly predicts "not injured" every time would achieve $98\%$ accuracy while failing to detect a single actual injury. This is known as the **Accuracy Paradox**.

$$Ratio = \frac{\sum \text{Non-Injured Records}}{\sum \text{Injured Records}}$$

Analyzing this ratio is the first step in deciding whether we need to implement specialized techniques such as **Synthetic Minority Over-sampling Technique (SMOTE)** or cost-sensitive learning (weighting the 'Injury' class more heavily).

```python
import pandas as pd

# Load the dataset
file_path = '../Dataset/week_approach_maskedID_timeseries.csv'
# Load the data
df = pd.read_csv(file_path)

# 1. Calculate the exact count of each class in the target variable ('injury')
# Why: We need absolute numbers to understand the mathematical skew of the dataset.
# In medical and sports datasets, the event of interest (injury) is usually rare.
class_counts = df['injury'].value_counts()

no_injury_count = class_counts[0]
injury_count = class_counts[1]
total_records = len(df)

# 2. Calculate the exact percentages for both classes
# Why: Percentages are necessary to explain the "Accuracy Paradox" in your report. 
# If a model predicts '0' every single time, its accuracy will equal the 'no_injury_percentage', 
# making it look like a highly accurate model while actually being useless for prediction.
no_injury_percentage = (no_injury_count / total_records) * 100
injury_percentage = (injury_count / total_records) * 100

# 3. Calculate the imbalance ratio
# Why: This ratio tells us how many negative samples exist for every positive sample. 
# This specific number will dictate which balancing technique to use later 
# (e.g., SMOTE, undersampling, or adjusting class weights in the loss function).
imbalance_ratio = no_injury_count / injury_count

# Print the results to be used in the EDA report
print("CLASS IMBALANCE ANALYSIS")
print(f"Total records: {total_records}")
print(f"No Injury (Class 0): {no_injury_count} ({no_injury_percentage:.2f}%)")
print(f"Injury (Class 1): {injury_count} ({injury_percentage:.2f}%)")
print(f"Imbalance Ratio: 1 injury for every {imbalance_ratio:.2f} non-injured records")
```
