**The aim of this file is to find the top 10 most injured and top 10 less injured athletes. Once this is achieved the goal is to try find differences in the trainings made by each of the groups. This can help find patterns that can lead athletes to suffer more injuries.** 

```python
import pandas as pd
import os
import matplotlib.pyplot as plt

# 1. Define the relative path to your file
# Navigate up from 'Codes' and down into 'Datasets'
file_path = '../Dataset/week_approach_maskedID_timeseries.csv'

# Load the data
df = pd.read_csv(file_path)

# Quick check to see if it loaded correctly
print(df.info())
```

```python

# 2. Group by athlete and add injuries
injuries_by_athlete = df.groupby('Athlete ID')['injury'].sum().reset_index()

# 3. Obtain 10 most injured athletes
top_10_more = injuries_by_athlete.sort_values(by='injury', ascending=False).head(10)

# 4. Obtain 10 less injured athletes
top_10_less = injuries_by_athlete.sort_values(by='injury', ascending=True).head(10)

# Show results
print("--- Top 10 Athletes  with more cummulated injuries ---")
print(top_10_more.to_string(index=False, header=["ID Athlete", "Total Injuries"]))

print("\n--- Top 10 Athletes  with less cummulated injuries ---")
print(top_10_less.to_string(index=False, header=["ID Athlete", "Total Injuries"]))
```

```python

# Define metrics we want to analyze
metrics = [
    'total kms', 
    'total hours alternative training', 
    'nr. rest days', 
    'avg exertion',
    'nr. sessions'
]

# 2. Group by athlete
athlete_stats = df.groupby('Athlete ID').agg({
    'injury': 'sum',
    **{m: 'mean' for m in metrics}
}).reset_index()

# 3. Calculate correlation with injury
correlation = athlete_stats.corr()['injury'].sort_values(ascending=False)

# 4. Divide in groups
top_10_more = athlete_stats.nlargest(10, 'injury')
top_10_less = athlete_stats.nsmallest(10, 'injury')

# 5. Compare groups
comparative = pd.DataFrame({
    'Metrics': metrics,
    'Average Top 10 more injured': [top_10_more[m].mean() for m in metrics],
    'Average Top 10 less injured': [top_10_less[m].mean() for m in metrics]
})

print("--- CORRELATION WITH INJURIES ---")
print(correlation)
print("\n--- PROFILE COMPARISON ---")
print(comparative.to_string(index=False))
```
